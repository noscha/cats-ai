```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import chisquare, ks_2samp


# ----------------------------
# Helper: chi-square on binned counts
# ----------------------------
def chi2_hist(a, b, bins):
    c1, _ = np.histogram(a, bins=bins)
    c2, _ = np.histogram(b, bins=bins)
    expected = c2 * (c1.sum() / c2.sum())
    return chisquare(c1, f_exp=expected)


# ----------------------------
# Load data: all columns
# ----------------------------
df1500_full = pd.read_csv("CarCrash/videos/Crash-1500/crash.csv")
df3000_full = pd.read_csv("CarCrash/videos/Normal/normal.csv")

# Select only numeric columns
numeric_cols = df1500_full.select_dtypes(include=[np.number]).columns
# Ensure both have the same columns (drop if missing in one)
numeric_cols = numeric_cols.intersection(df3000_full.select_dtypes(include=[np.number]).columns)

rng = np.random.default_rng(42)

# Store results in a list of dicts
results = []


for col in numeric_cols:
    # 1D arrays for this column
    df1500 = df1500_full[col].dropna().to_numpy()
    df3000 = df3000_full[col].dropna().to_numpy()

    if len(df1500) == 0 or len(df3000) == 0:
        continue  # skip empty columns

    # Shuffle and split 3000 into two halves
    idx = rng.permutation(len(df3000))
    split1 = df3000[idx[:len(idx)//2]]
    split2 = df3000[idx[len(idx)//2:]]

    # Common bins across all data for this column
    all_data = np.concatenate([df1500, split1, split2])
    bins = np.histogram_bin_edges(all_data, bins=30)

    # ----------------------------
    # Plot normalized histograms for this column
    # ----------------------------
    plt.figure()
    plt.hist(df1500, bins=bins, density=True, alpha=0.4, label="CSV 1500")
    plt.hist(split1, bins=bins, density=True, alpha=0.4, label="3000 half 1")
    plt.hist(split2, bins=bins, density=True, alpha=0.4, label="3000 half 2")
    plt.legend()
    plt.title(col)
    plt.xlabel("Value")
    plt.ylabel("Density")
    plt.show()

    # ----------------------------
    # Pairwise KS tests
    # ----------------------------
    ks_1500_s1 = ks_2samp(df1500, split1)
    ks_1500_s2 = ks_2samp(df1500, split2)
    ks_s1_s2 = ks_2samp(split1, split2)

    print(f"{col}: 1500 vs split1:", ks_1500_s1)
    print(f"{col}: 1500 vs split2:", ks_1500_s2)
    print(f"{col}: split1 vs split2:", ks_s1_s2)

    # ----------------------------
    # Chi-square tests on binned counts
    # ----------------------------
    chi_1500_s1 = chi2_hist(df1500, split1, bins)
    chi_1500_s2 = chi2_hist(df1500, split2, bins)
    chi_s1_s2 = chi2_hist(split1, split2, bins)

    print(f"{col}: Chi2 1500 vs split1:", chi_1500_s1)
    print(f"{col}: Chi2 1500 vs split2:", chi_1500_s2)
    print(f"{col}: Chi2 split1 vs split2:", chi_s1_s2)

    # ----------------------------
    # Collect results for this column
    # ----------------------------
    res = {
        "col": col,

        "ks_1500_vs_split1_stat": ks_1500_s1.statistic,
        "ks_1500_vs_split1_p": ks_1500_s1.pvalue,

        "ks_1500_vs_split2_stat": ks_1500_s2.statistic,
        "ks_1500_vs_split2_p": ks_1500_s2.pvalue,

        "ks_split1_vs_split2_stat": ks_s1_s2.statistic,
        "ks_split1_vs_split2_p": ks_s1_s2.pvalue,

        "chi_1500_vs_split1_stat": chi_1500_s1.statistic,
        "chi_1500_vs_split1_p": chi_1500_s1.pvalue,

        "chi_1500_vs_split2_stat": chi_1500_s2.statistic,
        "chi_1500_vs_split2_p": chi_1500_s2.pvalue,

        "chi_split1_vs_split2_stat": chi_s1_s2.statistic,
        "chi_split1_vs_split2_p": chi_s1_s2.pvalue,
    }

    results.append(res)


# ----------------------------
# Final results as DataFrame
# ----------------------------
if results:
    results_df = pd.DataFrame(results)
    print("\nResults summary:")
    print(results_df)

    # Optionally save to CSV
    # results_df.to_csv("crash_test_results.csv", index=False)
else:
    print("No numeric columns with data found.")
```
